4. Write a function that takes a number and returns **True** if it is even, else **False**.

```python
def check_even_odd(num):
    return num%2==0

print(check_even_odd(56))
print(check_even_odd(99))
```

5. Write a function with **default argument** that calculates simple interest
(Default rate = 5%).

```python
def simple_interest (p,t,r=5):
    return p*r*t / 100


print(simple_interest(20000,2,5))
print(simple_interest(20000,2,10))
    
```

6. Write a function that accepts **variable-length arguments** and returns their sum.

```python
def var_sum (*args):
    return sum(args)


print(var_sum(12,56,9,7,6,3))
print(var_sum(5))
```
