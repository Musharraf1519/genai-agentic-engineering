1. Write a program to find the **largest and smallest element** in a list.

```python
list1 = [1,2,4,9,22,56,12,43]


list2 = sorted(list1)

print(f"Largest element in {list1} is: {list2[-1]}")
print(f"Smallest element in {list1} is: {list2[0]}")
```

2. Write a program to **reverse a list** without using `reverse()`.

```python
list1 = [1,2,4,9,22,56,12,43]

print(list1[::-1])
```

3. Write a program to **remove duplicates** from a list.

```python
list1 = [1,2,4,9,22,56,12,43,12,4]

print(list(set(list1)))
```

4. Write a program to **count the frequency** of each element in a list.

```python
list1 = [1,2,4,9,22,56,12,43,12,4]

freq = {}

for i in list1:
    freq[i] = freq.get(i,0)+1
    
print(freq)
```

5. Write a program to **separate even and odd numbers** from a list.

```python
list1 = [1,2,4,9,22,56,12,43,12,4]
list_even=[]
list_odd = []

for i in list1:
    if i%2==0:
        list_even.append(i)
    else:
        list_odd.append(i)
        
print(list_even)
print(list_odd)
```

7. Write a program to **rotate a list to the right by K positions**.

```python
list1 = [1,2,4,9,22,56,12,43,12,45,54]

k =5

k = k % len(list1)
rotated_list = list1[-k:] + list1[:-k]
print(rotated_list)
```

8. Write a program to **check if a list is a palindrome**.

```python
lst1 = [1,2,1]
if lst1==lst1[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

10. Write a program to **find the maximum and minimum value** in a tuple.

```python
tuple1 = (1,2,3,4,5,6,1,4,1,4,5,1,6)
max_element = tuple1[0]
min_element = tuple1[0]

for i in tuple1:
    if max_element<i:
        max_element=i
        
    if min_element>i:
        min_element=i

# min_element = min(tuple)

print(max_element)
print(min_element)
```
