### **STRINGS IN PYTHON**

A **string** is an immutable sequence of Unicode characters. In Python, strings are treated as objects, which means they come packed with powerful built-in methods.

1. Key Properties

* Ordered: Each character has a specific position.
* Indexed: Access characters using numeric offsets.
* Iterable: You can loop through a string character by character.
* Immutable: Once created, the contents of a string object cannot be changed.
* Universal: Supports spaces, symbols, and emojis (Unicode).

*2. Immutability: The "No-Touch" Rule*
Because strings are immutable, you cannot modify them "in-place." Any operation that seems to change a string actually creates a brand-new string in memory.

**Note:**
Every string operation returns a **new string**.

````python
s = "hello"

# s[0] = "H" 
# This will raise a TypeError: 'str' object does not support item assignment


# The correct way: Create a new string
s = "H" + s[1:] 
print(s) # Output: "Hello"
````

*3. Indexing & Slicing*
Think of a string as a row of numbered boxes.
* Indexing: Access a single character.
    * s[0] is the first character.
    * s[-1] is the last character (negative indexing).

* Slicing: Extract a "substring" using the syntax [start : stop : step].
    * start: Included.
    * stop: Excluded (goes up to, but not including).
    * step: The interval between characters (default is 1).
* Reverse a String: Use slicing with step -1.

````python
s = "python"

s[0]     # 'p'
s[-1]    # 'n'
s[1:4:2]   # 'yth'
s[::-1]  # reverse string
````

*4. String Methods*<br>
**4.1 Case Conversion**

| Method         | Description                         | Example                 |
| -------------- | ----------------------------------- | ----------------------- |
| `upper()`      | Converts to uppercase               | `"abc".upper()` → `ABC` |
| `lower()`      | Converts to lowercase               | `"ABC".lower()` → `abc` |
| `title()`      | First letter of each word uppercase | `"hello world".title()` |
| `capitalize()` | Capitalizes first character only    | `"python".capitalize()` |

````python
text = "python IS fun"

print(text.upper())      # "PYTHON IS FUN"
print(text.lower())      # "python is fun"
print(text.title())      # "Python Is Fun" (Capitalizes every word)
print(text.capitalize()) # "Python is fun" (Only the first letter)
````

4.2 Validation Methods (Return Boolean)

| Method      | Checks             |
| ----------- | ------------------ |
| `isalpha()` | Only alphabets     |
| `isdigit()` | Only digits        |
| `isalnum()` | Alphabets + digits |
| `isspace()` | Only whitespace    |
| `islower()` | All lowercase      |
| `isupper()` | All uppercase      |
| `istitle()` | Title case         |

````python
"123".isdigit()    # True
print("abc".isalpha() )   # True
"a12".isalnum()    # True (Letters AND numbers)
"   ".isspace()    # True
"Python".istitle() # True
````

4.3 Searching & Counting

| Method         | Description                              |
| -------------- | ---------------------------------------- |
| `find()`       | Returns index, `-1` if not found         |
| `index()`      | Returns index, raises error if not found |
| `count()`      | Counts occurrences                       |
| `startswith()` | Checks prefix                            |
| `endswith()`   | Checks suffix                            |

````python
s = "python programming"

print(s.find("pro"))     # 7
print(s.count("o"))      # 2
print(s.startswith("py"))  # True
print(s.endswith("ing"))   # True
````

Replaces part of the string and returns a new one.

4.5 Splitting & Joining (The Power Couple)<br>
These are vital for data cleaning and transformation.

* split(sep): Breaks a string into a list. Default value for argument in split function is " "

* join(iterable): Glues a list back into a string.

````python
# Splitting
data = "apple,banana,cherry"
fruits = data.split(",") 
print(fruits) # ['apple', 'banana', 'cherry']

# Joining
words = ["Python", "is", "awesome"]
sentence = " ".join(words) 
print(sentence) # "Python is awesome"
````

4.6 Whitespace Handling

Cleaning of whitespace

````python
user_input = "   username123   "
print(user_input.strip())  # "username123" (Removes both sides)
user_input = "   username123   "
print(user_input.lstrip()) # "username123   " (Removes left)
user_input = "   username123   "
print(user_input.rstrip()) # "   username123" (Removes right)
````

### **Decorator**

A decorator is a design pattern that allows you to **modify or enhance the behavior of a function** without changing its source code. It wraps another function.

It relies on three Python pillars:
1. **First-Class Objects:** Functions can be assigned to variables.
2. **Nested Functions:** You can define a function inside another.
3. **Closures:** Inner functions remember the variables in their outer scope.

**2. Decorator Structure**

The `@decorator_name` syntax is just "syntactic sugar" for `func = decorator(func)`.

````python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Something is happening before the function is called.")
        result = func(*args, **kwargs) # Execute the original function
        print("Something is happening after the function is called.")
        return result
    return wrapper

@my_decorator
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Alice")
````

**3. Decorators with Arguments**

When you see a decorator like `@repeat(3)`, it’s actually a "decorator factory"—a function that returns a decorator.

````python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(n=3)
def greet():
    print("Hi!")

greet() # Prints "Hi!" three times
````

### **Iterators**


**1. Iterable vs. Iterator**

* **Iterable:** An object you can get an iterator from (e.g., List, Tuple, String). It has the `__iter__` method.
* **Iterator:** An object with a **state** that remembers where it is during iteration. It has the `__next__` method.

| Feature | Iterable | Iterator |
| --- | --- | --- |
| **Methods** | `__iter__()` | `__iter__()` and `__next__()` |
| **State** | Doesn't track position | Tracks current position |
| **Example** | `[1, 2, 3]` | `iter([1, 2, 3])` |

**2. The Iterator Protocol**

When you use a `for` loop, Python calls `iter()` on the object to get an iterator, then calls `next()` repeatedly until a `StopIteration` exception is raised.

````python
numbers = [1, 2]
it = iter(numbers)

print(next(it)) # 1
print(next(it)) # 2
# print(next(it)) # ❌ Raises StopIteration
````

### **Generators**

Generators are the "lazy" way to create iterators. Instead of `return` (which kills the function), they use **`yield`**.

* **Lazy Evaluation:** They produce items one at a time, only when asked.
* **Memory Efficient:** They don't store the entire list in memory.

**2. Execution Flow**

When a generator function is called, it returns a generator object **without** running the code. The code only runs when `next()` is called, stopping at each `yield`.

```python


```


```

````python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count  # Function "pauses" here
        count += 1

counter = count_up_to(3)
print(next(counter)) # 1
print(next(counter)) # 2
````

**3. Generator vs. List Comprehension**

````python
# List Comprehension (Eager - creates everything now)
my_list = [x**2 for x in range(1000000)] 

# Generator Expression (Lazy - creates on demand)
# Uses parentheses () instead of brackets []
my_gen = (x**2 for x in range(1000000)) 
````

### **10 Interview Questions (with concise answers)**<br>

1. Why are strings immutable in Python?<br>
**Answer:**
Strings are immutable to ensure memory efficiency, safety, and hashability. Any modification creates a new string object.

2. What happens internally when you modify a string in Python?<br>
**Answer:**
Python creates a **new string object** in memory; the original string remains unchanged.

3. Difference between indexing and slicing in strings?**
**Answer:**
Indexing accesses a single character, while slicing extracts a substring using `[start:stop:step]`.

4. Difference between `find()` and `index()` methods?**
**Answer:**
`find()` returns `-1` if the substring is not found, while `index()` raises a `ValueError`.

5. Why are `split()` and `join()` important in real applications?**
**Answer:**
They are essential for data cleaning and transformation—`split()` converts strings to lists, `join()` converts lists back to strings.

6. What is the best way to format strings in Python and why?**
**Answer:**
**f-strings** are best because they are faster, more readable, and evaluated at runtime.

7. What is a decorator in Python?<br>
**Answer:**
A decorator is a function that modifies or enhances another function’s behavior without changing its source code.

8. How does `@decorator` syntax work internally?<br>
**Answer:**
`@decorator` is syntactic sugar for `function = decorator(function)`.


9. What is the difference between an iterable and an iterator?<br>
**Answer:**
An iterable can produce an iterator, while an iterator maintains state and produces values using `__next__()`.

10. Why are generators more memory efficient than lists?<br>
**Answer:**
Generators produce values lazily using `yield`, storing only one value at a time instead of the entire sequence.
