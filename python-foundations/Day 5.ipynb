### **1. Functions in Python**

*1. A **function** is a reusable block of code that performs a specific task.*

**Why use functions?**
* Avoid code repetition
* Improve readability
* Easier debugging and testing

*1.1 Defining a Function (`def`)*

**Syntax**

```python
def function_name(parameters):
    statements
    return value
```

````python
def add(a, b):
    return a + b
````

*1.2 Calling a Function*

Calling a function means **executing** it.

````python
result = add(10, 20)
print(result)
````

*1.3 `return` Statement*

* Sends a value back to the caller
* Ends function execution
* A function can return **multiple values** (as a tuple)

````python
def calc(a, b):
    return a+b, a-b

x, y = calc(10, 5)

print(x,y)
````

*1.4 Function without `return`*<br>
Returns `None` implicitly.

````python
def greet():
    print("Hello")
    
print(greet())
````

### **2. Parameters vs Arguments**

*2.1 Parameters*<br>
* Variables listed in the function definition
* Act as **placeholders**

````python
def multiply(x, y):   # x, y → parameters
    return x * y
````

*2.2 Arguments*

* Actual values passed during function call

````python
multiply(3, 4)   # 3, 4 → arguments
````

*2.3 Difference (INTERVIEW TABLE)*

| Parameters          | Arguments                   |
| ------------------- | --------------------------- |
| Defined in function | Passed during function call |
| Receive values      | Provide values              |
| Local to function   | Exist at call time          |


*2.4 Types of Arguments (IMPORTANT)*

1. Positional Arguments<br>
Arguments are passed based on their **position** in the function call.

````python
def add(a, b):
    return a + b

add(2, 3)   # a=2, b=3
# ⚠ Order matters.
````

2. Keyword Arguments<br>
Arguments are passed using **parameter names**

````python
add(a=2, b=3)
# Order does NOT matter.
````

3. Default Arguments<br>
Parameters are assigned a **default value**, used if no argument is passed.


````python
def power(base, exp=2):
    return base ** exp

power(5)      # uses default exp = 2
power(5, 3)   # overrides default
# Default parameters must come after non-default parameters
````

4. `*args` — Variable-length Positional Arguments<br>
`*args` allows a function to accept **any number of positional arguments**.

````python
def total(*args):
    return sum(args)

total(1, 2, 3, 4)

# Collected as a tuple
````

5. `**kwargs` — Variable-length Keyword Arguments<br>
`**kwargs` allows a function to accept any number of keyword arguments.

````python
def details(**kwargs):
    print(kwargs)

details(name="Alice", age=25)
# Collected as a dictionary
````

### **3. Scope in Python (Local vs Global)**

Scope determines **where a variable can be accessed**.

*3.1 Local Scope*
* Variables defined inside a function
* Accessible only inside that function

````python
def test():
    xl = 10   # local variable
    print(xl)
    
# print(xl) #Outside access not allowed.
````

*3.2 Global Scope*
* Variables defined outside all functions
* Accessible everywhere (read-only by default)

````python
x = 100

def show():
    print(x)
````

*3.3 Modifying Global Variable (`global` keyword)*

**⚠ Use sparingly — bad practice in large programs.**

````python
count = 0

def increment():
    global count
    count += 1
````

*3.4 Local vs Global Example (TRAP QUESTION)*

````python
x = 5

def func():
    x = 10
    print(x)

func()
print(x)

# Local variable does not affect global variable.
````

### **4. LEGB Rule (VERY IMPORTANT)**

Python searches variables in this order:
1. **L**ocal
2. **E**nclosing
3. **G**lobal
4. **B**uilt-in

````python
x = 10

def outer():
    x = 20
    def inner():
        print(x)
    inner()

outer()
````

### **7. Recursive Functions**

A **recursive function** is a function that **calls itself** to solve a smaller version of the same problem.<br>
Every recursive function **must have**:
1. **Base case** → stops recursion
2. **Recursive case** → calls itself

*7.2 Basic Syntax*
```python
def function():
    if base_condition:
        return value
    return function(smaller_input)
```
7.3 Example: Factorial*

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
````

**Call Flow (n = 3)**

```python
factorial(3)
→ 3 * factorial(2)
→ 3 * 2 * factorial(1)
→ 3 * 2 * 1 * factorial(0)
→ 3 * 2 * 1 * 1
```

*7.4 Example: Fibonacci*

````python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
````

*7.5 Recursion vs Loop (INTERVIEW COMPARISON)*

| Recursion             | Loop               |
| --------------------- | ------------------ |
| Function calls itself | Repeats code block |
| Uses call stack       | No extra stack     |
| Elegant, readable     | More efficient     |
| Slower                | Faster             |


*7.6 Recursion Limit (IMPORTANT)*

Python limits recursion depth (~1000).

**⚠ Deep recursion → `RecursionError`**

````python
import sys
sys.getrecursionlimit()

````

*7.7 Common Recursion Mistakes*

* Missing base case
* Infinite recursion
* Excessive stack usage
* Using recursion where loop is better

*7.8 When to Use Recursion*

Use recursion when:
* Problem is naturally recursive (tree, graph)
* Divide-and-conquer algorithms
* Backtracking problems


### **8. Lambda Functions**

A **lambda function** is a **small anonymous function** written in **one line**.

**syntax**
```python
lambda arguments: expression
```

*8.2 Basic Example*

````python
add = lambda a, b: a + b
print(add(2, 3))

# Equivalent to:

def add(a, b):
    return a + b
````

*8.3 Characteristics of Lambda Functions*

* Anonymous (no name required)
* One expression only
* Implicit return
* Short-lived usage

*8.4 Lambda with Built-in Functions*

````python
### `map()`
nums = [1, 2, 3]
squares = list(map(lambda x: x*x, nums))

### `filter()`
even = list(filter(lambda x: x % 2 == 0, nums))

### `sorted()`
pairs = [(1, 3), (4, 1), (2, 2)]
sorted(pairs, key=lambda x: x[1])
````

*8.5 Lambda vs Normal Function*

| Lambda       | def           |
| ------------ | ------------- |
| One-line     | Multi-line    |
| Anonymous    | Named         |
| Simple logic | Complex logic |
| Short-lived  | Reusable      |


*8.6 Limitations of Lambda Functions*
* Only one expression
* No statements (loops, try/except)
* Hard to debug
* Lower readability if overused

*8.7 Common Lambda Mistakes*

* Writing complex logic
* Using lambda where `def` is clearer
* Forgetting it returns automatically


### **10 Interview Questions (with concise answers)**<br>
1. What is a function in Python? Why are functions used?

**Answer:**
A function is a reusable block of code that performs a specific task.
Functions reduce code duplication, improve readability, and make testing/debugging easier.

2. What is the difference between parameters and arguments?

**Answer:**
Parameters are variables defined in a function signature, while arguments are actual values passed during the function call.

3. What happens if a function does not have a `return` statement?

**Answer:**
The function returns `None` implicitly.

4. Can a Python function return multiple values? How?

**Answer:**
Yes. Python returns multiple values as a tuple.

```python
def calc(a, b):
    return a+b, a-b
```

5. Explain positional and keyword arguments with an example.

**Answer:**
Positional arguments are matched by position, while keyword arguments are matched by name.

```python
add(2, 3)        # positional
add(a=2, b=3)    # keyword
```

6. What is variable scope in Python?

**Answer:**
Scope defines where a variable can be accessed.
Python has local, enclosing, global, and built-in scopes.

7. What is the LEGB rule?

**Answer:**
Python resolves variable names in the order:
**Local → Enclosing → Global → Built-in**.

8. What is a recursive function? What are its two mandatory parts?

**Answer:**
A recursive function calls itself.
It must have:

1. A **base case** to stop recursion
2. A **recursive case** to reduce the problem

9. Why is recursion slower than loops in Python?

**Answer:**
Recursion uses the call stack and function overhead, while loops execute iteratively without extra stack usage.

10. What is a lambda function and when should it be used?**

**Answer:**
A lambda function is a one-line anonymous function used for short, simple operations, often with `map()`, `filter()`, or `sorted()`.

```python
lambda x: x * x
```
