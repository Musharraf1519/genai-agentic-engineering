### **OBJECT-ORIENTED PROGRAMMING (OOP) IN PYTHON**

*1. CLASS*<br>
A class is a user-defined data type that:
* groups data (variables) and behavior (methods)
* creates objects that share structure but hold different data


**Class Syntax**

```python
class ClassName:
    class_body
```

Rules:

* Class name follows **PascalCase**
* Code inside class runs **at class creation time**, not object creation time

In Python, **a class itself is an object** (of type `type`).

````python
class Student:
    pass

print(type(Student))   # <class 'type'>
````

*2. OBJECT*<br>
An **object is a runtime instance** of a class that:
* occupies memory
* has its own state (data)
* can access class behavior

````python
s1 = Student()
s2 = Student()
````

`s1` and `s2`:
* same class
* different memory locations

````python
print(id(s1), id(s2))
````

*3. CONSTRUCTOR*<br>

A constructor in Python is a special method that runs automatically when a new object is created from a class.<br>
Its job is to initialize (set up) the object’s data

When you write:

```python
obj = MyClass()
```

Python does **two things**:

1. Creates an empty object in memory
2. Calls the constructor to initialize it

The constructor is the method named:

```python
__init__
```

**Basic constructor example**

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
s1 = Student("Rahul", 20)
s2 = Student("Anita", 22)
````

What happens internally:

* `s1` gets its own `name` and `age`
* `s2` gets its own `name` and `age`
* Same class, different object data

````python
## Constructor without parameters
class Demo:
    def __init__(self):
        print("Object created")

d = Demo()
````

````python
## Constructor with default values
class Account:
    def __init__(self, balance=0):
        self.balance = balance

a1 = Account()
a2 = Account(1000)
````

Multiple constructors?<br>
Python **does NOT support method overloading**, including constructors.<br>

This is not invalid:

```python
def __init__(self):
    pass

def __init__(self, x):
    pass
```
The second one overwrites the first.

Correct way → use default arguments:

```python
def __init__(self, x=0):
    self.x = x
```

What happens if you don’t write a constructor?

Python provides a **default constructor** automatically.

```python
class A:
    pass

a = A()   # works
```

But:

* You cannot initialize data
* Object will be empty


**Constructor Rules**

* Name must be `__init__`
* First parameter is **always the object reference**
* Can have default arguments
* Can be overloaded only via defaults (Python has no constructor overloading)


*4. `self` KEYWORD*<br>
`self` is **a reference to the current object**.

Without `self`, Python has no way to know **which object** is calling the method.

````python
class Test:
    def show(self):
        print(self)
        
t = Test()
t.show()
````

*5. INSTANCE VARIABLES*

Variables **unique to each object**.

````python
class Student:
    def __init__(self, name):
        self.name = name
        
        
s1 = Student("A")
s2 = Student("B")

# `s1.name ≠ s2.name`

print(s1.name)
print(s2.name)
````

Instance Variable are dynamic in nature and can be added outside the constructor.

````python
s1.marks = 90

# This variable exists only for s1

print(s1.marks)
# print(s2.marks) #Does not exist
````

*6. CLASS VARIABLES*<br>
Variables **shared by all objects** of a class.

````python
class Student:
    school = "ABC School"

### Access Rules
Student.school
s1 = Student()
s1.school
````

````python
# *Shadowing Trap (INTERVIEW FAVORITE)*
s1.school = "XYZ"


# This creates an **instance variable**, does NOT change class variable.
print(Student.school)   # "ABC School"
print(s1.school)        # "XYZ"
````

*7. METHODS IN DETAIL*<br>
**7.1 Instance Methods**

* A method that works on **object (instance) data**
* Takes **`self`** as the first parameter
* Can access and modify **instance variables**

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


acc = BankAccount(1000)
acc.deposit(500)

print(acc.balance)   # 1500
````

**7.2 Class Method**
* A method that works on **class-level data**
* Uses **`@classmethod`**
* Takes **`cls`** as the first parameter
* Can access and modify **class variables**

````python
class BankAccount:
    bank_name = "SBI"

    @classmethod
    def change_bank(cls, name):
        cls.bank_name = name


BankAccount.change_bank("HDFC")

print(BankAccount.bank_name)  # HDFC
````

**7.3 Static Method**
* A method that does **not depend on instance or class data**
* Uses **`@staticmethod`**
* Takes **no automatic parameters**

````python
class BankAccount:
    @staticmethod
    def is_valid_amount(amount):
        return amount > 0


print(BankAccount.is_valid_amount(500))   # True
print(BankAccount.is_valid_amount(-10))   # False
````

**When to use Which Method**

Use **instance method** when the method:

* needs to access or modify object data
* depends on instance variables
* behaves per object


**Example use cases:**

* deposit / withdraw money
* update user profile
* calculate result based on object data


Use **class method** when the method:

* needs to access or modify class-level data
* applies to all objects
* is used as an alternative constructor


**Example use cases:**

* change shared configuration
* create object from file / string / JSON
* maintain object count


Use **static method** when the method:

* does not depend on instance or class data
* is logically related to the class
* is a utility/helper function


**Example use cases:**

* validation logic
* helper calculations
* formatting utilities

**Key Difference**

| Method Type | Parameter | Access           |
| ----------- | --------- | ---------------- |
| Instance    | `self`    | Instance + Class |
| Class       | `cls`     | Class only       |
| Static      | None      | Neither          |


*8. Pillars of OOPs Concept*

**8.1 Encapsulation in Python**

**Encapsulation** means bundling data with the methods that operate on it and controlling direct access to that data so the object cannot be put into an invalid state.

In simple words:
*“Don’t allow anyone to freely change important data.”*


**Why Encapsulation is needed*

Without encapsulation:

```python
class Account:
    def __init__(self, balance):
        self.balance = balance
```

```python
acc = Account(1000)
acc.balance = -5000    # invalid, but allowed
```

Problem:

* Data can be changed directly
* No validation
* Object state becomes unreliable




**Encapsulation in Practice**

````python
class Account:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance

# Usage:
acc = Account(1000)

acc.deposit(500)
acc.withdraw(300)

print(acc.get_balance())   # 1200


acc.__balance = -10000
print(acc.get_balance())   # still 1200
````

**What Encapsulation achieved here**

* `__balance` cannot be modified directly
* All changes go through methods with rules
* Object always stays valid
* Internal data is **protected**

This is **real encapsulation**.



**How Python implements Encapsulation**

Python uses **conventions**, not strict enforcement:

1. Public:
```python
self.balance
```
**Accessible everywhere.**

2. Protected (convention):
```python
self._balance
```
**internal use, don’t touch directly**.

3. Private (name mangling)
```python
self.__balance
```
**Internally becomes `_ClassName__balance`.**


**Note**
Encapsulation is NOT security.<br>
You can still access private data if you try hard:
```python
acc._Account__balance
```

Python trusts developers — encapsulation is about **design discipline**, not hiding secrets.

*8.2. INHERITANCE*<br>
**Inheritance** means one class (child) acquiring properties and methods of another class (parent).

In simple words:
*Reuse existing code instead of writing it again.*

````python


## Basic Inheritance Example
class Account:
    def deposit(self, amount):
        print("Deposited:", amount)

class SavingsAccount(Account):
    pass

s = SavingsAccount()
s.deposit(1000)

# * `SavingsAccount` **inherits** from `Account`
# * `deposit()` is reused
# * No duplication



## Parent and Child Relationship
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    pass

c = Child()
c.show()   # Parent method


# Child automatically gets parent methods.




## Inheritance with Constructor
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")

c = Child()


### `super()`

# * Calls parent class method
# * Used to reuse parent initialization
````

**Types of Inheritance in Python**

1. Single Inheritance
```python
class Child(Parent):
    pass
```
2. Multilevel Inheritance

```python
class A:
    pass

class B(A):
    pass

class C(B):
    pass
```

3. Multiple Inheritance

```python
class A:
    pass

class B:
    pass

class C(A, B):
    pass
```

Python allows multiple inheritance.


**When to use Inheritance**<br>
Use inheritance when:

* Child and parent share common behavior
* There is a clear IS-A relationship
* Code reuse is required



**When NOT to use Inheritance**<br>
Avoid inheritance when:

* Relationship is HAS-A (use composition)
* Only small functionality is shared
* Parent class changes often

*8.3. ABSTRACTION*<br>
**Abstraction** means showing only what an object can do and hiding how it does it.

In simple words:
*User knows the function, not the internal logic.*

Without abstraction, users depend on implementation details.
```python
engine.start_fuel_pump()
engine.check_sensors()
engine.ignite()
```

User should not care about this complexity.Instead:
```python
engine.start()
```
That’s abstraction.

**How Python implements Abstraction**

Python uses **Abstract Base Classes (ABC)** from the `abc` module.

Key ideas:

* Abstract class **cannot be instantiated**
* Abstract methods **must be implemented by child classes**
* Forces a **common structure**

**Abstract Class Example**

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass
````

This class:

* defines **what** must be done
* does not define **how**

**Child Class Implementation**

````python
class Car(Vehicle):
    def start(self):
        print("Car engine started")
        
c = Car()
c.start()
````

**What happens if method is not implemented**

```python
class Bike(Vehicle):
    pass

b = Bike()   # ERROR
```

Error:

```
TypeError: Can't instantiate abstract class
```

This **forces correctness**.

**Real-World Example**

````python
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using card")


class UpiPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI")

p = CardPayment()
p.pay(500)
````

User only knows:

```python
pay(amount)
```

**When to use Abstraction**<br>
Use abstraction when:

* You want to **define a contract**
* Multiple implementations are possible
* You want to **force structure**
* You want loose coupling

*8.4 Polymorphism*<br>
**Polymorphism** means **one interface, multiple behaviors**.

In simple words:
*Same method name, but different behavior depending on the object.*

Without polymorphism, you write separate logic for each type.<br>
With polymorphism:
* same method call
* different objects
* correct behavior chosen automatically

This makes code **flexible and extensible**.

1. Polymorphism via Method Overriding (Most Common)<br>
Child class provides its **own version** of a parent method.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")
        

a1 = Dog()
a2 = Cat()

a1.sound()   # Dog barks
a2.sound()   # Cat meows
````

* Same method name `sound()`
* Different behavior
* Decision made at **runtime**

2. Polymorphism using Same Function, Different Objects

````python
def make_sound(animal):
    animal.sound()
    
make_sound(Dog())
make_sound(Cat())
````

The function doesn’t care **what the object is**,only that it has a `sound()` method.<br>
This is **true polymorphism**.

3. Polymorphism with Built-in Functions

````python
print(len("Python"))     # 6
print(len([1, 2, 3]))    # 3
print(len((4, 5)))       # 2


# Same function `len()`
# Different data types
# Different behavior
````

4. Polymorphism via Operator Overloading.<br>
Operators behave differently based on operands.

````python
print(10 + 5)        # 15
print("Py" + "thon") # Python
````

Behind the scenes:

* `+` calls different methods
* based on object type

**When to use Polymorphism**<br>
Use polymorphism when:
* you want to write **generic code**
* behavior depends on object type
* you want to avoid `if/else` chains
* future extensions are expected

*9. Dunder Methods in Python*<br>
**Dunder methods** are special methods in Python that start and end with **double underscores** (`__method__`).<br>
They allow your **objects to behave like built-in types**.


These methods let Python understand how your object should behave when you use:

* `print()`
* `len()`
* `+`
* `==`
* iteration (`for`)
* indexing (`[]`)

Instead of writing logic manually every time.

1. `__init__` (Constructor)<br>
Called when an object is created.

````python
class Person:
    def __init__(self, name):
        self.name = name

p = Person("Rahul")
print(p.name)
````

2. `__str__` (String representation)<br>
Controls what `print(object)` shows.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

p = Person("Rahul")
print(p)   # Rahul
````

3. `__len__` (Length)<br>
Used by `len()`.

````python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

t = Team(["A", "B", "C"])
print(len(t))   # 3
````

4. `__add__` (Operator overloading)<br>
Defines behavior of `+`.

````python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

a = Number(10)
b = Number(20)

print(a + b)
````

5. `__eq__` (Equality `==`)

Controls comparison using `==`.

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks

s1 = Student(90)
s2 = Student(90)

print(s1 == s2) 

6. `__getitem__` (Indexing)<br>
Allows object[index].

````python
class Data:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index]

d = Data([10, 20, 30])
print(d[1])
````

7. `__iter__` and `__next__` (Iteration)<br>
Make object iterable.

````python
class Count:
    def __init__(self, n):
        self.n = n
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i < self.n:
            self.i += 1
            return self.i
        raise StopIteration
````

**Commonly Used Dunder Methods**

| Method        | Purpose         |
| ------------- | --------------- |
| `__init__`    | Object creation |
| `__str__`     | Print object    |
| `__len__`     | Length          |
| `__add__`     | `+`             |
| `__eq__`      | `==`            |
| `__getitem__` | Indexing        |
| `__iter__`    | Iteration       |
| `__next__`    | Next value      |

### **10 Interview Questions (with concise answers)**<br>

1. What is a class in Python?<br>
**Answer:**
A class is a user-defined blueprint that groups data (variables) and behavior (methods) and is itself an object of type `type`.

2. What is an object? How is it different from a class?<br>
**Answer:**
An object is a runtime instance of a class that occupies memory and holds its own state, while a class is just a blueprint.

3. What is the purpose of the `__init__` method?<br>
**Answer:**
`__init__` is a constructor that initializes object data and is automatically called when an object is created.

4. What happens if you don’t define a constructor in a class?<br>
**Answer:**
Python provides a default constructor, allowing object creation but without initializing instance variables.

5. What is `self` in Python?<br>
**Answer:**
`self` is a reference to the current object, used to access instance variables and methods inside the class.

6. Difference between instance variables and class variables?<br>
**Answer:**
Instance variables are unique to each object, while class variables are shared among all objects of the class.

7. Explain instance, class, and static methods.<br>
**Answer:**

* Instance method → works on object data (`self`)
* Class method → works on class data (`cls`)
* Static method → utility logic, no access to class or object data

8. What is encapsulation in Python? How is it achieved?<br>
**Answer:**
Encapsulation protects object data by restricting direct access, typically using private variables (`__var`) and controlled methods.

9. What is polymorphism in Python?<br>
**Answer:**
Polymorphism allows the same method name to exhibit different behavior depending on the object calling it, resolved at runtime.

10. What are dunder methods? Give two examples.<br>
**Answer:**
Dunder methods are special methods with double underscores that define object behavior.
Examples:

* `__str__` → controls `print()` output
* `__add__` → defines `+` operator behavior
