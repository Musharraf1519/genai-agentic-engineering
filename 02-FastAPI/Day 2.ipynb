### **Section 8: Path Parameters**

We are now entering Parameters & Data Handling. This is where your API stops being a static "Hello World" and starts becoming dynamic by accepting information from the user.<br><br>
A **Path Parameter** is a variable part of your URL path. It allows you to use the same "route" to fetch different specific pieces of data based on what the user types in the address bar.<br>
1. The Syntax:<br>
To create a path parameter, you wrap a variable name in curly braces `{ }` inside the path of your decorator. You then declare that same variable as an argument in your function.<br>
**Example Code:**

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/items/{item_id}")
async def read_item(item_id: int):
    return {"item_id": item_id, "message": f"You are looking at item {item_id}"}
````

2. Technical Breakdown ⚙️

* **`{item_id}`**: This tells FastAPI: "Whatever the user puts in this spot of the URL, capture it and call it `item_id`."
* **`item_id: int`**: This is where the **Python Type Hints** we learned in Section 4 become powerful.
* If a user visits `/items/42`, FastAPI sees that `42` can be an integer. It converts it from a string to a Python `int` automatically and passes it to your function.
* If a user visits `/items/foo`, FastAPI sees that "foo" is **not** an integer. It will automatically stop the request and send back a "422 Unprocessable Entity" error before your function even runs.

3. Order Matters<br>
When creating multiple routes, FastAPI matches them from **top to bottom**. If you have a fixed path and a dynamic path that look similar, the fixed path must come first.<br>
**Example of the correct order:**

````python
@app.get("/users/me")
async def read_user_me():
    return {"user_id": "the current user"}

@app.get("/users/{user_id}")
async def read_user(user_id: str):
    return {"user_id": user_id}

# If you put `{user_id}` first, the API would think "me" is just another ID and would never reach the specific `/users/me` logic.
````

### **Section 9: Query Parameters**

When you declare function parameters that are **not** part of the path, FastAPI automatically interprets them as "Query Parameters." These are the key-value pairs that appear after a `?` in a URL.

1. The Syntax

````python
@app.get("/items/")
async def read_items(skip: int = 0, limit: int = 10):
    return {"skip": skip, "limit": limit}
````

* **URL Example**: `http://127.0.0.1:8000/items/?skip=20&limit=50`
* **Technical Result**:
* `skip` becomes `20`.
* `limit` becomes `50`.

2. Default Values and Optionality<br>
Query parameters are great because they can have **default values**. In the example above, if the user just visits `/items/` without any extra text, the API will automatically use `0` for skip and `10` for limit.<br>
If you want a parameter to be truly optional (meaning it can be `None`), you can use Python's `Optional` type:

````python
from typing import Optional

@app.get("/search/")
async def search(q: Optional[str] = None):
    if q:
        return {"query": q}
    return {"message": "No query provided"}
````

To ensure you have a complete, "spoonfed" understanding, we will build a single script that demonstrates exactly how **Path Parameters** and **Query Parameters** work together in a real, functional API.<br>
We will use a **Store Inventory** scenario. We need to find a specific item by its ID (Path Parameter) and optionally filter the search results by name or price (Query Parameters).
<br>The Complete Working Example<br>
Create a file (e.g., `inventory_api.py`) and paste the following code:

````python
from fastapi import FastAPI
from typing import Optional

app = FastAPI()

# 1. Path Parameter Example
# This route identifies a specific item in our database by its ID.
@app.get("/items/{item_id}")
async def get_item_by_id(item_id: int):
    """
    'item_id' is a Path Parameter. It is MANDATORY.
    If you go to /items/5, item_id becomes 5.
    If you go to /items/abc, FastAPI returns an error because 'abc' is not an 'int'.
    """
    return {
        "item_id": item_id, 
        "location": "Warehouse A",
        "status": "In Stock"
    }

# 2. Query Parameter Example
# This route handles a search. Notice 'q' and 'short_description' are not in the URL path.
@app.get("/items/")
async def search_items(q: str, short_description: bool = False):
    """
    'q' is a REQUIRED Query Parameter because it has no default value.
    'short_description' is an OPTIONAL Query Parameter with a default of False.
    URL Example: /items/?q=hammer&short_description=true
    """
    results = {"search_query": q}
    
    if short_description:
        results.update({"description": "A high-quality tool for construction."})
    else:
        results.update({"description": "Full details are hidden. Set short_description=True to see more."})
        
    return results

# 3. Combined Path and Query Parameters
@app.get("/users/{user_id}/orders")
async def get_user_orders(user_id: int, limit: Optional[int] = None):
    """
    'user_id' is a Path Parameter (Mandatory).
    'limit' is an Optional Query Parameter.
    URL Example: /users/101/orders?limit=5
    """
    orders = [f"Order {i}" for i in range(1, 11)] # Simulating 10 orders
    
    if limit:
        return {"user_id": user_id, "orders": orders[:limit]}
    
    return {"user_id": user_id, "orders": orders}
````

2. **Test the Path Parameter:**

### **Section 10: The Request Body**

In the previous sections, we sent data to the server using the URL (Path and Query parameters). However, when you need to send complex data—like a user's registration details, a long blog post, or a product's metadata—putting everything in the URL is insecure and limited.<br>
This is where the **Request Body** comes in. A Request Body is data sent by the client to the API as a JSON object, hidden from the URL.
1. The Role of Pydantic Models:<br>
To receive a Request Body in FastAPI, you must define a **Schema**. This is a Python class that inherits from Pydantic's `BaseModel`. This model acts as a "contract": it defines exactly what fields the data must have and what types those fields must be.<br>
**Example Model:**

````python
from pydantic import BaseModel
from typing import Optional

class Item(BaseModel):
    name: str
    description: Optional[str] = None
    price: float
    tax: Optional[float] = None
````

2. Creating the Route<br>
When you add this model as a type hint in your function, FastAPI realizes that this data should not be taken from the URL, but should be read from the **Request Body**.

````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Item(BaseModel):
    name: str
    price: float


@app.post("/items/")
async def create_item(item: Item):
    # 'item' is now a Python object. 
    # You can access data using item.name or item.price
    return {"message": f"Item {item.name} created", "total_price": item.price}
````

3. Technical Breakdown: What FastAPI Does<br>
When a client sends a **POST** request to `/items/` with a JSON body, FastAPI performs these steps automatically:
* **Read** the request body as JSON.
* **Convert** the JSON data into Python types (e.g., converting `"19.99"` to a float `19.99`).
* **Validate** the data: If `name` is missing or `price` is a string like "expensive", FastAPI sends an error response immediately.
* **Give you the object**: It hands you the `item` variable already populated with data.

4. A Complete Working Example: Combining Everything<br>
In real applications, you often use **Path**, **Query**, and **Body** parameters all at once. Let's see how they coexist in one function.

````python
from fastapi import FastAPI
from pydantic import BaseModel
from typing import Optional

app = FastAPI()

class User(BaseModel):
    username: str
    email: str
    bio: Optional[str] = None

# Combining Path, Query, and Body
@app.put("/users/{user_id}")
async def update_user(
    user_id: int,           # Path Parameter
    user: User,             # Request Body
    importance: int = 1     # Query Parameter
):
    results = {
        "user_id": user_id, 
        "user_data": user, 
        "importance": importance
    }
    return results
````
