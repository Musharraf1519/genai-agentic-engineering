### **The Comprehensive Guide to Web Communication & REST APIs**

*1. The Client-Server Model*

At its core, the internet functions on a request-response pattern. This is known as the **Client-Server Architecture**. It is a distributed application structure that partitions tasks or workloads between the providers of a resource or service (servers) and service requesters (clients).

**The Participants**

* **The Client:** The "requester." It is any device or software (like a web browser, a mobile app, or a command-line tool) that triggers a request for data or a specific action.
* **The Server:** The "provider." It is a powerful computer or software process that sits waiting for requests. When one arrives, it processes the logic, interacts with a database if necessary, and sends a response back.

**The 4-Step Transaction Flow**

1. **Initiation:** The client sends an HTTP request to a specific address (URL).
2. **Reception:** The server receives the request and parses the metadata (headers and path).
3. **Processing:** The server executes code (e.g., "Find user with ID 10 in the database").
4. **Completion:** The server sends an HTTP response containing a status code and the requested data.

**Example:**

```example
Browser → Google server → Search results
```

````python
import requests

# 1. Client creates the request
url = "https://api.github.com/users/octocat"

# 2. Client sends the request to the Server
response = requests.get(url)

# 3. Server sends back the response
print(f"Status Code: {response.status_code}")
print(f"Server Type: {response.headers['Server']}")
````

*2. HTTP: The Language of the Web*

**HTTP (HyperText Transfer Protocol)** is the standardized set of rules that defines how the client and server must format their messages so they can understand each other.It defines:
* how requests are sent
* how responses are returned
* data formats
* status handling

**Key Technical Characteristics:**

* **Statelessness:** The server does not "remember" previous requests. Every request is treated as brand new. To the server, a request sent one second ago is unrelated to a request sent now.
* **TCP/IP Foundation:** HTTP usually runs on top of Transmission Control Protocol (TCP), which ensures that data packets arrive in the correct order without errors.
* **Media Independence:** While we mostly send JSON today, HTTP can carry images, video, HTML, or plain text.

````python
# Request 1: We tell the server who we are (hypothetical)
response1 = requests.get("https://httpbin.org/get", params={"user": "Rahul"})

# Request 2: We send a new request without the parameter
response2 = requests.get("https://httpbin.org/get")

# The server has no memory of "Rahul" from the previous line.
# Every request must contain all information necessary to understand it.
print(response1.json()['args'])
print(response2.json()['args']) # Output: {} (Empty)
````

*3. Anatomy of an HTTP Transaction*

To understand APIs, you must be able to "read" the raw messages being sent back and forth.

A. The HTTP Request (Client → Server)

A request consists of four main parts:
1. **URL (Uniform Resource Locator):** The address of the resource (e.g., `https://api.example.com/products`).
2. **Method:** The verb indicating the action (GET, POST, etc.).
3. **Headers:** Key-value pairs providing metadata (e.g., `Content-Type: application/json` or `Authorization: Bearer <token>`).
4. **Body:** The data being sent to the server (used in POST, PUT, and PATCH).

B. The HTTP Response (Server → Client)

A response consists of three main parts:

1. **Status Code:** A 3-digit number indicating if the request was successful or why it failed.
2. **Headers:** Information about the server and the data being returned.
3. **Body:** The payload (usually the data you requested in JSON format).

Here is a breakdown of what a full exchange looks like in code.

````python


# The Request (Sent by you)

headers = {'User-Agent': 'MyPythonApp/1.0', 'Accept': 'application/json'}
url = "https://httpbin.org/get"

response = requests.get(url, headers=headers)
# The Response (Received by you)

print("--- Response Headers ---")
print(response.headers['Content-Type']) # application/json

print("\n--- Response Body ---")
print(response.text) # The raw string data returned by server
````

*4. HTTP Methods*

In a RESTful system, the method tells the server what **action** to perform on a resource.

| Method | CRUD Action | Description |
| --- | --- | --- |
| **GET** | **Read** | Retrieves data. It should never change the state of the server. |
| **POST** | **Create** | Submits new data to the server to create a new resource. |
| **PUT** | **Update** | Replaces an existing resource entirely with new data. |
| **PATCH** | **Update** | Updates only specific fields of an existing resource. |
| **DELETE** | **Delete** | Removes a specific resource from the server. |

We will use a "Dummy API" to simulate a database of users.

````python


# GET (Read)
# Fetching user ID 1
res = requests.get("https://jsonplaceholder.typicode.com/users/1")
print(res.json()["name"]) # Output: Leanne 

# POST (Create)
# Creating a new user
new_user = {"name": "Alice", "email": "alice@example.com"}
res = requests.post("https://jsonplaceholder.typicode.com/users", json=new_user)
print(res.status_code) # 201 (Created)

# PUT (Update)
# Replacing user 1 with new data
updated_data = {"name": "Alice Updated", "email": "alice_new@example.com"}
res = requests.put("https://jsonplaceholder.typicode.com/users/1", json=updated_data)
print(res.json()) # Shows the updated object

# DELETE (Remove)
# Removing user 1
res = requests.delete("https://jsonplaceholder.typicode.com/users/1")
print(res.status_code) # 200 or 204 (Success)
````

*5. HTTP Status Codes (The Feedback)*

Status codes are grouped into four categories based on the first digit:

🟢 2xx: Success

* **200 OK:** The request was successful.
* **201 Created:** A new resource was successfully created (common after POST).
* **204 No Content:** Success, but there is no data to return (common after DELETE).

🟡 3xx: Redirection

* **301 Moved Permanently:** The resource is now at a different URL.
* **304 Not Modified:** Tells the client to use its cached version of the data.

🟠 4xx: Client Error (You did something wrong)

* **400 Bad Request:** The server didn't understand the request (syntax error).
* **401 Unauthorized:** You need to log in or provide an API key.
* **403 Forbidden:** You are logged in but don't have permission for this specific resource.
* **404 Not Found:** The resource does not exist at that URL.

🔴 5xx: Server Error (The server failed)

* **500 Internal Server Error:** A generic "something went wrong on our end" message.
* **503 Service Unavailable:** The server is overloaded or down for maintenance.

A good developer writes code to handle different status codes.

````python
def check_url(url):
    res = requests.get(url)
    
    if res.status_code == 200:
        print("Success!")
    elif res.status_code == 404:
        print("Error: The resource was not found.")
    elif res.status_code == 500:
        print("Error: The server is having a meltdown.")
    else:
        print(f"Received status: {res.status_code}")

check_url("https://google.com/non-existent-page") # Triggers 404 logic
````

*6. JSON: The Data Format*

**JSON (JavaScript Object Notation)** is the industry standard for API data exchange. It is text-based and easy for both humans and machines to parse.

JSON Syntax Rules:

1. Data is in **name/value pairs** (e.g., `"name": "Rahul"`).
2. Data is separated by **commas**.
3. Curly braces `{}` hold **objects**.
4. Square brackets `[]` hold **arrays**.
5. Keys must be wrapped in **double quotes**.

Data Type Mapping

| JSON Type | Python Equivalent | Example |
| --- | --- | --- |
| String | `str` | `"Hello World"` |
| Number | `int` or `float` | `42` or `3.14` |
| Object | `dict` | `{"id": 1}` |
| Array | `list` | `[1, 2, 3]` |
| Boolean | `bool` | `true` / `false` |
| null | `None` | `null` |

JSON Example

```json
{
  "name": "Rahul",
  "age": 25,
  "skills": ["Python", "SQL"],
  "active": true
}

API data usually arrives as a string. We must convert it into a "Dictionary" (in Python) to use it.

````python
import json

# This is a raw string (looks like JSON)
raw_data = '{"id": 505, "status": "active", "tags": ["admin", "editor"]}'

# Convert string to Python Dictionary
data = json.loads(raw_data)

# Now we can access it using keys
print(data["id"])        # 505
print(data["tags"][0])   # admin
````

*7. REST Architecture Principles*

**REST (Representational State Transfer)** is not a language or a tool, but a set of architectural constraints. If an API follows these rules, it is called **RESTful**.

1. **Uniform Interface:** You should use the same naming conventions for all URLs (e.g., plural nouns like `/users` instead of `/getUser`).
2. **Client-Server:** The UI (Client) and the Data (Server) are separate. You can replace the UI without touching the Server.
3. **Stateless:** No client context is stored on the server between requests.
4. **Cacheable:** Responses must define themselves as cacheable or not to improve performance.
5. **Layered System:** A client cannot tell whether it is connected directly to the end server or to an intermediary (like a load balancer).

### Example of a RESTful URL Structure:

* `GET /books` — Get all books.
* `POST /books` — Add a new book.
* `GET /books/101` — Get details for book ID 101.
* `DELETE /books/101` — Remove book ID 101.

*8. Putting It All Together: A Real-World Flow*

Imagine you are using a weather app on your phone.

1. **The Request:** Your app (Client) sends a request:
* **URL:** `https://api.weather.com/v1/forecast/london`
* **Method:** `GET`
* **Headers:** `Authorization: API_KEY_123`


2. **The Processing:** The Weather Server receives the request, validates the API key, and queries its database for London's weather.
3. **The Response:** The Server sends back:
* **Status Code:** `200 OK`
* **Body:**
```json
{
  "city": "London",
  "temperature": 18,
  "unit": "Celsius",
  "description": "Partly Cloudy"
}


Example to perform a GET request to fetch the current temperature and weather conditions for a specific location (London).

````python
import requests

# 1. THE REQUEST: Defining our target and parameters
# Latitude and Longitude for London
url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 51.5074,
    "longitude": 0.1278,
    "current_weather": "true"
}

print(f"--- Step 1: Client sends GET request to {url} ---")
response = requests.get(url, params=params)

# 2. THE PROCESSING: (Happens on the Open-Meteo server)
# The server looks up the coordinates, finds the latest sensor data, 
# and packages it into a JSON object.

# 3. THE RESPONSE: Receiving the data
print(f"--- Step 3: Server returns Status Code: {response.status_code} ---")

if response.status_code == 200:
    # 4. DATA PARSING: Converting JSON to a Python Dictionary
    data = response.json()
    
    current = data["current_weather"]
    temp = current["temperature"]
    wind = current["windspeed"]
    
    print("\n--- Final Result (Displayed to User) ---")
    print(f"Location: London")
    print(f"Temperature: {temp}°C")
    print(f"Wind Speed: {wind} km/h")
else:
    print("Failed to retrieve data.")
````
