7. Write a program to **count unique vowels in a string using a set**.

```python
s="python"
str_set = set(s)
unique_vowel =0

for i in str_set:
    if i.lower() in ['a','e','i','o','u']:
        unique_vowel +=1
        
print(unique_vowel)
```

11. Write a program to **count the frequency of characters in a string**.

```python
s = "python"

freq = {}

for i in s:
    freq[i] = freq.get(i,0)+1
    
print(freq)
```

12. Write a program to **count the frequency of words in a sentence**.

```python
s = "Python is the best Programming Language in The world."
words = s.split()
freq = {}

for i in words:
    freq[i.lower()] = freq.get(i.lower(),0)+1
    
print(freq)
```

13. Write a program to **find the key with the maximum value in a dictionary**.

```python
d1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

max_value = 0
max_key =""
for k,v in d1.items():
    if v>max_value:
        max_value = v
        max_key = k
print(max_key)
```

16. Write a program to **remove keys with duplicate values**.

```python
d1 = {'a': 1, 'b': 2, 'c': 2, 'd': 3, 'e': 1}

value_count = {}
for v in d1.values():
    value_count[v] = value_count.get(v, 0) + 1

result = {}
for k, v in d1.items():
    if value_count[v] == 1:
        result[k] = v

print(result)
```
