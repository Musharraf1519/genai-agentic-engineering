### **EXCEPTION HANDLING IN PYTHON**

**Exception Handling** is a mechanism to handle runtime errors so that the program **does not crash abruptly**. By "catching" these errors, you ensure the program can continue execution gracefully or shut down safely, preventing data loss or a poor user experience.


*2. What is an Exception?*<br>
An **exception** is a signal that something went wrong during the execution of a program. Unlike **syntax errors** (which prevent code from running at all), exceptions occur while the code is running. When an error occurs, Python creates an exception object. If not "caught," the program halts and prints a traceback.

**Example**

````python
# x = 10 / 0
````

*3. Why Exception Handling is Needed*<br>
In production environments, a single error shouldn't crash an entire server.

* **Without handling:** The program terminates immediately, and any "cleanup" code (like closing a database) or the remaining code is skipped.
* **With handling:** The error is intercepted, alternate logic (like logging the error or asking for new input) runs, and the program stays alive.


*4. The `try...except` Block*<br>

The `try` block lets you test a block of code for errors. The `except` block lets you handle the error.<br>
**Basic Syntax**

```python
try:
    # risky code (code that might fail)
except:
    # handling code (executes ONLY if an error occurs)

```
**Example**

````python
try:
    x = 10 /10
except:
    print("Error occurred")
````

*4. Catching Specific Exceptions*

Using a "bare" `except:` is generally discouraged because it catches *everything*, including system exits and keyboard interrupts (). By specifying the error type, you ensure you are only responding to errors you actually expect.


````python
try:
    x = int("abc")
except ValueError:
    print("Invalid conversion")
````

*Multiple Blocks*

````python
try:
    x = int("10")
    y = 10 / 0
except ValueError:
    print("Conversion error")
except ZeroDivisionError:
    print("Division by zero")
````

*Multiple Exceptions in One Block*
Use a tuple to catch various errors in one go. Using `as e` allows you to inspect the error message generated by Python.

````python
try:
    x = 10 / 0
except (ZeroDivisionError, ValueError) as e:
    print("Error:", e)
````

*6. The `else` Block*

The `else` block is the "success path." It runs **only if no exception occurs** in the `try` block. This is useful for code that should only execute if the "risky" part was successful, keeping the `try` block as small as possible.

````python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Error")
else:
    print("Division successful")
````

*7. The `finally` Block**

The `finally` block is the "cleanup crew." It **always executes**, regardless of whether an exception was raised or caught. It is primarily used for releasing external resources, such as closing file streams or database connections.

````python
try:
    x = 10 / 2
except:
    print("Error")
finally:
    print("Execution finished")
````

*8. Full Structure & Execution Order*

When all blocks are combined, they follow a strict logical flow:

1. **`try`**: Run the code.
2. **`except`**: Run only if an error occurred.
3. **`else`**: Run only if no error occurred.
4. **`finally`**: Run no matter what happened above.

```python
try:
    # Attempt this
    pass
except ExceptionType:
    # Handle if failed
    pass
else:
    # Do this if successful
    pass
finally:
    # Do this last, always
    pass

```

*9. Raising Exceptions Manually**

Sometimes, a program is syntactically correct, but the logic is wrong (e.g., a negative age). You can use the `raise` keyword to force an exception to occur.

````python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
````

*10. Custom Exceptions**

For complex applications, you can create your own exception types by inheriting from the built-in `Exception` class. This makes your code more readable and easier to debug.

````python
class InvalidAgeError(Exception):
    """Exception raised for errors in the input age."""
    pass

age = -1

if age < 0:
    raise InvalidAgeError("Invalid age")
````

*11. Common Built-in Exceptions Reference*

| Exception | Cause |
| --- | --- |
| `ZeroDivisionError` | Occurs when the denominator of a division is zero. |
| `ValueError` | Occurs when a function receives an argument of right type but inappropriate value. |
| `TypeError` | Occurs when an operation is applied to an object of inappropriate type. |
| `IndexError` | Occurs when trying to access an index that is out of range. |
| `KeyError` | Occurs when a dictionary key is not found. |
| `FileNotFoundError` | Occurs when a file or directory is requested but doesn't exist. |
| `AttributeError` | Occurs when a reference or assignment fails for a specific attribute. |

*12. Exception Hierarchy*

Python exceptions are organized in a tree structure. Catching a parent class (like `Exception`) will also catch all of its children (like `ValueError`).

```python
BaseException
 ├── SystemExit (Interpreter shutdown)
 ├── KeyboardInterrupt (User hits Ctrl+C)
 └── Exception (User-defined/Standard errors)
     ├── ValueError
     ├── TypeError
     ├── IndexError
     └── ...

```

### **FILE HANDLING IN PYTHON**

File handling is a crucial part of any application, allowing data to persist after a program ends. In Python, files are treated as either **text** or **binary** sequences.

File handling lets you:

* create files
* read data
* write data
* append data


When you open a file, you must specify a **mode**. This tells the operating system whether you intend to read, write, or modify the file.

```python
file = open("data.txt", "mode")

```

**Common Modes**

| Mode | Meaning | Technical Note |
| --- | --- | --- |
| `"r"` | **Read** | Default. File must exist or it raises `FileNotFoundError`. |
| `"w"` | **Write** | Creates file if missing; **truncates (wipes)** existing content. |
| `"a"` | **Append** | Adds data to the end without deleting existing content. |
| `"x"` | **Exclusive** | Fails if the file already exists (prevents accidental overwrites). |
| `"r+"` | **Read/Write** | Allows updating a file without closing it. |



**2. Reading Files**

Reading involves moving data from the disk into the computer's RAM.

* **`read()`**: Best for small files where you need the whole string.
* **`readline()`**: Efficient for very large files; it only loads one line into memory at a time.
* **`readlines()`**: Returns a list of strings, which is handy for indexing specific lines.

````python
# 2.1 Read Entire File
f = open("data.txt", "r")
content = f.read()
print(content)
f.close()

# 2.2 Read First N Characters
f.read(10)

# 2.3 Read Line by Line
while True:
    line = f.readline()
    if not line: break
    print(line)

# 2.4 Read All Lines as List
lines = f.readlines()
print(lines)
````

*3. Writing to Files*

* **Writing (`"w"`)**: Think of this as "starting fresh." Even if you don't write anything, opening in `"w"` mode clears the file.
* **Appending (`"a"`)**: This moves the **File Pointer** to the very end before writing, keeping your history intact.

**Examples:**

````python
# 3.1 Write (Overwrites)
f = open("data.txt", "w")
f.write("Hello World")
f.close()

# 3.2 Write Multiple Lines
lines = ["A\n", "B\n", "C\n"]
f = open("data.txt", "w")
f.writelines(lines)
f.close()

# 4. Appending
f = open("data.txt", "a")
f.write("\nNew line added")
f.close()
````

*4. Using the `with` Statement (Best Practice)*

The `with` statement creates a **Context Manager**. It ensures that the file is closed automatically as soon as the block of code finishes, even if an error occurs. This prevents **memory leaks** and file corruption.

```python
with open("data.txt", "r") as f:
    print(f.read())

```

````python
with open("data.txt", "r") as f:
    print(f.read())
````

Why use `with`:

* no manual `close()`
* safer
* handles exceptions

*6. File Pointer (`seek` & `tell`)*

**Current position**

````python
f.tell()
````

Move pointer

````python
f.seek(5)   # move to position 5
f.seek(0)   # start of file
````

*7. Example — Read and Write Together*

````python
with open("data.txt", "w") as f:
    f.write("Python\n")

with open("data.txt", "a") as f:
    f.write("File Handling\n")

with open("data.txt", "r") as f:
    print(f.read())
````

*8. Important Notes*

* Always close files (or use `with`)
* `"w"` deletes old data
* `"a"` preserves old data
* Reading large files → use line iteration

### **CSV File Handling in Python**

CSV = **Comma Separated Values**
Used to store tabular data (rows & columns).

Example CSV file:

```
name,age,city
Rahul,25,Delhi
Anita,30,Mumbai
```

Python uses the built-in **`csv` module**.

*1. Reading CSV File*

````python
import csv

with open("data.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
        
# Each row → list.
````

*2. Reading CSV as Dictionary*

````python
import csv

with open("data.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
        
# Column names become keys.
````

*3. Writing CSV File*

````python
import csv

data = [
    ["name", "age", "city"],
    ["Rahul", 25, "Delhi"],
    ["Anita", 30, "Mumbai"]
]

with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
````

*4. Writing CSV using DictWriter*

````python
import csv

data = [
    {"name": "Rahul", "age": 25},
    {"name": "Anita", "age": 30}
]

with open("data.csv", "w", newline="") as file:
    fieldnames = ["name", "age"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)
````

### **JSON File Handling in Python**

JSON = **JavaScript Object Notation**
Used for APIs, configs, data exchange.

Example JSON:

````python
{
  "name": "Rahul",
  "age": 25
}

# Python uses the built-in **`json` module**.
````

*1. Reading JSON File*

````python
import json

with open("data.json", "r") as file:
    data = json.load(file)

print(data)
print(data["name"])


# `json.load()` → file → Python dict.
````

*2. Writing JSON File*

````python
import json

data = {
    "name": "Rahul",
    "age": 25,
    "city": "Delhi"
}

with open("data.json", "w") as file:
    json.dump(data, file)
````

*3. Pretty Printing JSON*

```python
json.dump(data, file, indent=4)
```

Adds formatting for readability.

*4. JSON String Handling*

Dict → JSON string

```python
json_string = json.dumps(data)
```

JSON string → Dict

```python
data = json.loads(json_string)
```

**CSV vs JSON (Quick Difference)**

| CSV             | JSON                  |
| --------------- | --------------------- |
| Tabular data    | Nested data           |
| Comma separated | Key-value             |
| Less flexible   | More flexible         |
| Smaller size    | Larger but structured |

### **Word File Handling in Python (.docx)**

Python uses the **`python-docx`** library.

Install:

```bash
pip install python-docx
```

*1. Creating / Writing a Word File*

````python
from docx import Document

doc = Document()

doc.add_heading("Report", level=1)
doc.add_paragraph("This is a sample paragraph.")

doc.save("sample.docx")

# This creates a Word document.
````

*2. Reading a Word File*

````python
from docx import Document

doc = Document("sample.docx")

for para in doc.paragraphs:
    print(para.text)
````

*3. Adding Multiple Paragraphs*

````python
doc = Document()

doc.add_paragraph("Line 1")
doc.add_paragraph("Line 2")
doc.add_paragraph("Line 3")

doc.save("multi.docx")
````

*4. Adding Tables*

````python
doc = Document()

table = doc.add_table(rows=2, cols=2)

table.rows[0].cells[0].text = "Name"
table.rows[0].cells[1].text = "Age"

table.rows[1].cells[0].text = "Rahul"
table.rows[1].cells[1].text = "25"

doc.save("table.docx")
````

### **PDF File Handling in Python**

Common libraries:

* `PyPDF2` → read PDFs
* `reportlab` → create PDFs

Reading PDF (PyPDF2)

Install:

```bash
pip install PyPDF2
```

*1. Read PDF Text*

````python
import PyPDF2

with open("sample.pdf", "rb") as file:
    reader = PyPDF2.PdfReader(file)

    for page in reader.pages:
        print(page.extract_text())
        
# Reads text from all pages.
````

*2. Number of Pages*

````python
print(len(reader.pages))
````

*3. Writing / Creating PDF (reportlab)*

Install:

```bash
pip install reportlab
```

*4. Create Simple PDF*

````python
from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet

doc = SimpleDocTemplate("sample.pdf")

styles = getSampleStyleSheet()
content = []

content.append(Paragraph("Hello PDF", styles["Normal"]))

doc.build(content)


# Creates a basic PDF.
````

**2. Multiple Lines PDF**

````python
content.append(Paragraph("Line 1", styles["Normal"]))
content.append(Paragraph("Line 2", styles["Normal"]))
````

### **10 Interview Questions (with concise answers)**<br>

1. What is Exception Handling in Python?<br>
**Answer:**
Exception handling is a mechanism to manage runtime errors so the program doesn’t terminate abruptly. It uses constructs like `try`, `except`, `else`, and `finally` to detect and handle errors gracefully.

2. What is the difference between a Syntax Error and an Exception?<br>
**Answer:**
* **Syntax Error:** Occurs before execution due to invalid code structure.
* **Exception:** Occurs during runtime when syntactically correct code fails (e.g., division by zero).

5. What is the role of the `finally` block?
**Answer:**
`finally` always executes, regardless of whether an exception occurs. It is used for cleanup tasks like closing files or releasing database connections.

Using the `raise` keyword:

```python
if age < 0:
    raise ValueError("Age cannot be negative")
```

class InvalidAgeError(Exception):
