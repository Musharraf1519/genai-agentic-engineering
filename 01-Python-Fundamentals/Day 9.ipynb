### **Input Validation**

**Input validation** is the process of ensuring that a program operates on clean, correct, and useful data. It acts as a "security guard" for your code, blocking malicious or accidental entries that could cause crashes or security vulnerabilities.

*The Theory Behind Validation*

* **Prevent Errors:** Blocks "garbage" data (e.g., text where a number belongs).
* **Enforce Rules:** Ensures data makes sense logically (e.g., an age cannot be negative).
* **Data Integrity:** Guarantees that the variables in your program always hold valid states.

**Example — Without Validation**

Without checks, Python assumes the user will always follow instructions.

````python
age = int(input("Enter age: "))
print(age + 10)
````

**Risk:** If the user enters `"abc"`, the `int()` function fails, raising a `ValueError` and crashing the program immediately.

### **Approaches to Validation**

*A. Manual Type Checking*

Using string methods like `.isdigit()` to check if the input can safely be converted.

````python
age_input = input("Enter age: ")

if age_input.isdigit():
    age = int(age_input)
    print(age + 10)
else:
    print("Invalid input")
````

*B. Range & Logic Validation*

Ensuring the data fits within realistic boundaries.

````python
age = int(input("Enter age: "))

if 0 <= age <= 120:
    print("Valid age")
else:
    print("Invalid age range")
````

*C. Exception-Based Validation (Pythonic Approach)*

Instead of checking before doing, we "try" to do it and handle the failure. This is often more robust.

````python
try:
    age = int(input("Enter age: "))
except ValueError:
    print("Please enter a number")
````

### **2. Logging Basics**

**Logging** is a professional way of tracking what happens when your software runs. While `print()` statements are useful during local development, they are temporary and hard to manage in large systems.

**Why Logging Wins Over Print()**

* **Persistence:** Logs can be saved to files to review crashes that happened weeks ago.
* **Severity:** You can categorize messages (e.g., a "Warning" doesn't stop the program, but an "Error" might).
* **Filtering:** You can choose to see only "Critical" errors in production to keep logs clean.


**Logging Levels**

| Level | Purpose |
| --- | --- |
| **DEBUG** | Detailed information, typically of interest only when diagnosing problems. |
| **INFO** | Confirmation that things are working as expected (e.g., "User logged in"). |
| **WARNING** | An indication that something unexpected happened, but the app is still running. |
| **ERROR** | A serious problem; the program hasn't been able to perform a specific task. |
| **CRITICAL** | A major failure; the program itself may be unable to continue running. |


**Implementation**

Basic Setup

````python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
logging.warning("Low balance")
logging.error("Transaction failed")
````

**Logging to a Permanent File**
By adding a `format`, you can include timestamps automatically.

````python
import logging

logging.basicConfig(
    filename="test_log.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")

import os
print(os.path.abspath("test_log.log"))
````

### **3. Debugging Stack Traces**

A **stack trace** (or traceback) is a report showing the active function calls at the moment an unhandled exception occurred. It is a "map" that leads you back to the root cause of a crash.

### **Anatomy of a Stack Trace**

Python traces follow a specific structure:

1. **Call Sequence:** Every function currently "open" in memory.
2. **Location:** The filename and line number for every step.
3. **The Culprit:** The final line containing the Exception Type and specific error message.

**Reading Strategy: Bottom → Top**

* **Bottom:** Find out **what** happened (e.g., `ZeroDivisionError`).
* **Middle:** Find out **where** it happened (the line of code).
* **Top:** Find out **how** you got there (the chain of functions).

````python
# def divide(a, b):
#     return a / b

# divide(10, 0)
````

**The Output:**

```text
Traceback (most recent call last):
  File "app.py", line 4, in <module>
    divide(10, 0)
  File "app.py", line 2, in divide
    return a / b
ZeroDivisionError: division by zero
```

### **Advanced Traceback Handling**

**Manual Extraction**
Useful when you want to log the error details without actually letting the program crash.

````python
import traceback

try:
    10 / 0
except:
    traceback.print_exc() # Prints the trace but continues execution
````

**Logging the Full Trace**
In production, we use `exc_info=True` to save the entire stack trace into our log file for later debugging.

````python
import logging

try:
    10 / 0
except Exception as e:
    logging.error("Error occurred", exc_info=True)
````

### **Garbage Collection in Python**

**Garbage Collection (GC)** is the process of **automatically freeing memory** by deleting objects that are no longer in use.<br>
In simple words:<br>
When an object is not needed anymore, Python removes it from memory.<br>
This prevents memory leaks.

**Why Garbage Collection is needed**<br>
When you create objects:
```python
a = [1, 2, 3]
```
Memory is allocated.
If Python never removed unused objects → memory would keep filling → program/system crash.

GC ensures:

* efficient memory usage
* automatic cleanup


**Two Mechanisms Used by Python**
Python uses **two memory management techniques**:
1. Reference Counting (primary)
2. Cyclic Garbage Collection (secondary)

1. Reference Counting<br>
Every object has a **reference count** → number of variables pointing to it.

Example:

```python
a = [1, 2, 3]
```

Ref count = 1

```python
b = a
```

Ref count = 2

If you delete one reference:

```python
del a
```

Ref count = 1

When it becomes **0**:

```python
del b
```

Object is immediately removed from memory.

2. Cyclic Garbage Collector

To handle circular references, Python has a **cycle detector**. It:
* scans objects
* detects reference loops
* removes unreachable cycles

This runs periodically.

# GC Module (Manual Control)

Python provides the `gc` module.

Enable / Disable GC

````python
import gc

gc.disable()
gc.enable()
````

Force Garbage Collection

````python
gc.collect()


# Returns number of objects collected.
````

Check Garbage Collector Stats

````python
gc.get_count()

# Shows collection counters for generations.
````

**Objects are collected when:**
* Reference count = 0
* No reachable references exist
* Circular references are detected


**Important Notes**

* Python GC is automatic
* Developers rarely manage memory manually
* GC handles most cases efficiently
* Cycles + large objects → GC becomes important
