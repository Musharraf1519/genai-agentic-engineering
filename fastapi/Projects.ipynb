Here are three **Level 1** problems. They focus purely on the fundamentals: **Path Parameters**, **Query Parameters**, and **Pydantic Models**.

**Problem 1: The "Digital Greeting Card"**<br>
**Goal:** Practice Path and Query Parameters.<br>
**The Mission:**<br>
Create a single GET route called `/greet/{name}`.

1. **Path Parameter:** It should take the `name` from the URL.
2. **Query Parameter:** It should take an optional `message` (string).
3. **The Logic:** * If a message is provided, return: `"Hello [name], [message]"`
* If no message is provided, return a default: `"Hello [name], hope you have a great day!"`



**Problem 2: The "Fruit Stand" Inventory**<br>
**Goal:** Practice Pydantic Models and POST requests.<br>
**The Mission:**<br>
Create a POST route called `/fruits/` that allows a user to "add" a fruit to a list.
1. **The Model:** Create a Pydantic class called `Fruit`. It should have:
* `name` (string)
* `grams` (float)
* `is_ripe` (boolean)

2. **The Logic:**
* The route should accept this `Fruit` model in the **Request Body**.
* The API should return a "receipt" that says: `"Received [grams]g of [name]."`
* **Constraint:** If the fruit is not ripe, add a note to the response saying: `"Warning: This fruit needs more time!"`


**Problem 3: The "Simple Calculator"**<br>
**Goal:** Practice Data Validation and Error Handling.<br>
**The Mission:**<br>
Create a GET route called `/calculate`.<br>
1. **Query Parameters:** Accept three parameters: `x` (float), `y` (float), and `operation` (string).
2. **The Logic:**
* Support four operations: `add`, `subtract`, `multiply`, and `divide`.
* Return the result in a JSON object like: `{"result": 10.5}`.
3. **The Challenge:**
* If the user tries to `divide` by `0`, raise a **FastAPI HTTPException** with a `400` status code and the message: `"Cannot divide by zero, math is crying."`
* If the user types an operation that isn't one of the four listed above, return a `404` error.

Now, let's step up the difficulty just a bit. These **Level 2** problems introduce **State** (remembering things), **List handling**, and **Dependency Injection basics**.

Problem 4: The "Digital Guestbook"<br>
**Goal:** Practice persistent state (in-memory) and List responses.<br>
**The Mission:** Create a simple guestbook where people can leave their names and a short note.

2. **The POST Route (`/sign`):** * Accepts a Pydantic model `Entry` with `name` and `note`.
* Adds the entry to the `guestbook` list.
3. **The GET Route (`/entries`):** * Returns the *entire* list of entries.
4. **The Challenge:** Add a Query Parameter to the GET route called `limit`. If the user hits `/entries?limit=2`, only return the last 2 people who signed.


Problem 5: The "Secure Vault" (Basic Dependency)<br>
**Goal:** Practice your first **Dependency**.<br>
**The Mission:** You have a secret piece of data: `"The treasure is buried under the palm tree."`
1. **The Dependency:** Create a function called `verify_header`.
* It should look for a **Header** called `X-Secret-Key`. (Hint: Use `from fastapi import Header`).
* If the header is exactly `"OpenSesame"`, it returns `True`.
* If it’s anything else, it raises an `HTTPException(401)`.
2. **The Route:** Create a GET route `/vault`.
* Use `Depends(verify_header)` to protect it.
* If the dependency passes, return the secret message.



Problem 6: The "Order Tracker" (Nested Models)<br>
**Goal:** Practice Pydantic models inside other models.
**The Mission:** In the real world, data is "nested" (objects inside objects).
1. **The Models:** * Create a `Product` model (`name`, `price`).
* Create an `Order` model that contains a **List** of `Product` objects and a `total_price`.
2. **The POST Route (`/checkout`):**
* Accepts an `Order` object.
3. **The Logic:** * Your code should loop through the list of products, add up their prices, and check if the sum matches the `total_price` sent by the user.
* If the math doesn't match, return a `400` error saying: `"Price mismatch! Your math is wrong."`
