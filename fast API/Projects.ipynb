### **Project A: The "Pulse" Health-Tech API**

**The Scenario:** You are building the backend for a fitness wearable. The API must handle high-velocity data (heart rate logs) while keeping sensitive user data secure.<br>
1. Technical Requirements:<br>
* **User Management:** Users must be able to sign up with an email and password. Use **Pydantic** to ensure the email is valid and the password is at least 8 characters long.
* **Authentication:** Protect the health data routes. Only the owner of the data (or an admin) should be able to view it.
* **Data Logging (The Heartbeat):**
* Create a POST route that accepts a **Request Body** containing `heart_rate` (int) and `timestamp` (datetime).
* This route should use **Path Parameters** to identify the user ID.


* **Performance:** Use **Background Tasks** to calculate the user's "Daily Average Heart Rate" and save it to a separate table so the main request remains fast.
* **Database:** Use **SQLAlchemy** to store users and their heart rate logs in a relational database.

2. The Challenge Logic:
* Implement a **Dependency** that checks if the user's heart rate is above 180. If it is, the response should include a special "Warning" flag in the JSON.
* Create a **Response Model** for the User Profile that hides the `hashed_password` and only shows the `username` and `join_date`.

### **Project B: The "FlashCart" E-Commerce Engine**

**The Scenario:** You are building the engine for a lightning-fast flash sale site. Thousands of users will try to buy limited stock at the same time.<br>
1. Technical Requirements:<br>
* **Inventory System:**
* Create a `Product` model with `name`, `price`, and `stock_count`.
* The `stock_count` cannot be negative. Use **Pydantic validation** or **SQLAlchemy constraints** to enforce this.

* **Advanced Filtering (Query Parameters):**
* The GET `/products` route must support pagination (`skip` and `limit`).
* It must also allow filtering by a price range (e.g., `min_price` and `max_price`).


* **The Order Process:**
* When an order is placed, use a **Database Transaction** to ensure that if the stock is 10, and two people buy at the exact same microsecond, the stock doesn't end up at -1.


* **Error Handling:**<br>
* If a user tries to buy an item that is out of stock, raise a custom **HTTPException** with a `400 Bad Request` and a detailed message: "Insufficient stock for product [Name]".

2. The Challenge Logic:<br>
* **Async/Await Mastery:** Create an endpoint that "simulates" checking a third-party shipping provider (like FedEx or UPS). Use `asyncio.sleep` to simulate the network delay without blocking other users from browsing products.<br>
* **Security:** Create an "Admin" role. Only admins should have access to the DELETE route for products. Use **Dependency Injection** to verify the user's role from their **JWT Token**.<br>
