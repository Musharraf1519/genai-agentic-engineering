###  **Python requests module**

*1. Installation and Importing*

The library is not part of the Python Standard Library. You must install it via your terminal or command prompt:

```bash
pip install requests

```

Once installed, you must import it at the top of your `.py` file:

````python
import requests
````

*2. The GET Request*

The `GET` method is used to retrieve data from a specified resource.

````python
# The basic syntax
response = requests.get('https://jsonplaceholder.typicode.com/posts/1')

# The 'response' variable is an object of the Response class
print(type(response)) 
````

Essential Response Attributes

When a server responds, the `response` object contains several key attributes:

* **`status_code`**: An integer representing the HTTP status (e.g., **200** for success, **404** for not found).
* **`url`**: The final URL of the request (useful if there were redirects).
* **`encoding`**: The encoding used to decode `response.text`.
* **`headers`**: A dictionary-like object of the response headers.

*3. Accessing Response Body*

There are three primary ways to access the data returned by the server:

A. Raw Text

Used for HTML or plain text.

````python
print(response.text) # Returns a Unicode string
````

B. JSON Data

Used when interacting with APIs. This converts the JSON string into a Python dictionary or list.

````python
data = response.json()
print(data['title']) # Accessing a key in the dictionary
````

C. Binary Content

Used for non-textual files like images, PDFs, or zip files.

````python
print(response.content) # Returns bytes
````

*4. Query String Parameters*

To send data in the URL (e.g., `?id=1&user=admin`), use the `params` argument. Requests will automatically format the URL for you.

````python
query_params = {'userId': 1}
response = requests.get('https://jsonplaceholder.typicode.com/posts', params=query_params)

print(response.url) 
# Output: https://jsonplaceholder.typicode.com/posts?userId=1
````

*5. The POST Request*

The `POST` method is used to send data to a server to create or update a resource. The data is sent in the body of the request, not the URL.


````python
payload = {
    'title': 'foo',
    'body': 'bar',
    'userId': 1
}

# Use the 'json' parameter to send a dictionary as a JSON object
response = requests.post('https://jsonplaceholder.typicode.com/posts', json=payload)

print(response.status_code) # Usually 201 for 'Created'
````

*6. Request Headers*

Headers allow you to pass additional information with your request, such as authentication tokens or the "User-Agent" (identifying your software).

````python
custom_headers = {
    'Content-Type': 'application/json',
    'Authorization': 'Bearer YOUR_TOKEN_HERE'
}

response = requests.get('https://api.example.com/data', headers=custom_headers)
````

7. Error Handling

Network requests can fail due to DNS issues, timeouts, or server errors. You should use `raise_for_status()` to catch HTTP errors (4xx or 5xx).

````python
try:
    response = requests.get('https://api.example.com/data', timeout=3)
    # If the response was unsuccessful, this raises an HTTPError
    response.raise_for_status()
except requests.exceptions.HTTPError as http_err:
    print(f"HTTP error occurred: {http_err}")
except requests.exceptions.ConnectionError as conn_err:
    print(f"Error connecting: {conn_err}")
except requests.exceptions.Timeout as time_err:
    print(f"The request timed out: {time_err}")
````

8. State Persistence (Sessions)

By default, every call to `requests.get()` or `requests.post()` is independent. If you need to keep cookies or maintain a persistent connection to a server, use a `Session` object.

````python
# Create a session object
session = requests.Session()

# All requests made with 'session' will share cookies
session.get('https://httpbin.org/cookies/set/sessioncookie/12345')
response = session.get('https://httpbin.org/cookies')

print(response.json())
````
